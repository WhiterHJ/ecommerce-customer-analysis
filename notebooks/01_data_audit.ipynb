## 缺失值检查

```python
missing_summary = pd.DataFrame({
    "missing_count": df_raw.isna().sum(),
    "missing_percent": (df_raw.isna().mean() * 100).round(2)
})

missing_summary.sort_values(
    "missing_count",
    ascending=False
)
```

```python
missing_relationship = pd.crosstab(
    df_raw["CustomerID"].isna(),
    df_raw["Description"].isna(),
    rownames=["CustomerID缺失"],
    colnames=["Description缺失"],
    margins=True
)

missing_relationship
```

```python
both_missing_count = (
    df_raw["CustomerID"].isna()
    & df_raw["Description"].isna()
).sum()

print("客户编号和商品描述同时缺失的行数：", both_missing_count)
```

## 取消订单检查

```python
invoice_text = (
    df_raw["InvoiceNo"]
    .astype("string")
    .str.strip()
)

is_cancelled = invoice_text.str.upper().str.startswith(
    "C",
    na=False
)

is_negative_quantity = df_raw["Quantity"] < 0
```

```python
cancellation_check = pd.Series({
    "订单编号以C开头的记录": is_cancelled.sum(),
    "C开头且数量为负": (
        is_cancelled & is_negative_quantity
    ).sum(),
    "C开头但数量非负": (
        is_cancelled & ~is_negative_quantity
    ).sum(),
    "数量为负但订单号不以C开头": (
        ~is_cancelled & is_negative_quantity
    ).sum()
})

cancellation_check
```

```python
negative_without_c = df_raw.loc[
    ~is_cancelled & is_negative_quantity,
    [
        "InvoiceNo",
        "StockCode",
        "Description",
        "Quantity",
        "UnitPrice",
        "CustomerID",
        "Country"
    ]
]

negative_without_c.head(20)
```

```python
negative_without_c[
    ["StockCode", "Description"]
].value_counts(dropna=False).head(20)
```

```python
non_positive_price = df_raw.loc[
    df_raw["UnitPrice"] <= 0,
    [
        "InvoiceNo",
        "StockCode",
        "Description",
        "Quantity",
        "UnitPrice",
        "CustomerID",
        "Country"
    ]
]

non_positive_price.head(20)
```

```python
df_raw.loc[
    df_raw["UnitPrice"] < 0,
    [
        "InvoiceNo",
        "StockCode",
        "Description",
        "Quantity",
        "UnitPrice",
        "CustomerID",
        "Country"
    ]
]
```

```python
stock_text = (
    df_raw["StockCode"]
    .astype("string")
    .str.strip()
)

is_standard_stock = stock_text.str.fullmatch(
    r"\d{5}[A-Za-z]?",
    na=False
)

print("标准形式商品编码数量：", is_standard_stock.sum())
print("非标准形式商品编码数量：", (~is_standard_stock).sum())
```

```python
special_stock_summary = (
    df_raw.loc[
        ~is_standard_stock,
        ["StockCode", "Description"]
    ]
    .value_counts(dropna=False)
    .head(30)
)

special_stock_summary
```

```python
country_summary = (
    df_raw
    .groupby("Country", dropna=False)
    .agg(
        record_count=("InvoiceNo", "size"),
        invoice_count=("InvoiceNo", "nunique"),
        customer_count=("CustomerID", "nunique")
    )
    .sort_values(
        "record_count",
        ascending=False
    )
)

country_summary.head(10)
```
